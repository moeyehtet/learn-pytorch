### Creating a Tensor with `requires_grad`

If you create a variable which will need to be modified by calculating gradients in the optimization, use `requires_grad=True` when creating the tensor.

```python
x = torch.ones(5, requires_grad=True)
x
```

## Autograd

```python
x = torch.randn(3, requires_grad=True)
x
```

You will see `grad_fn` attribute which is used for calculating gradients for backpropagation if you create the tensor with `requires_grad=True`.

```python
y = x + 2
y
```

```python
z = y * y * 2
z
```

```python
z = z.mean()
z
```

To calculate gradients:

```python
z.backward() # calculate dz/dx
```

```python
x.grad # gradients of x tensor
```

**Note**: This works now because `z.mean()` results in a scalar. If it is a vector, we need to pass in a vector the same size as our original vector.

```python
z = y * y *2

v = torch.tensor([0.1, 1.0, 0.001], dtype=torch.float32)
z.backward(v)
x.grad
```

Preventing from tracking the history and calculating the `grad_fn` attribute. 
- For example, when we want to update our weights, the gradient calculation should not be part of the operation.

```python
# 1st option
# x.requires_grad_(False)
x.requires_grad_(False) # will modify the tensor in placd
x
```

```python
# 2nd option
y = x.detach() # will create a new tensor without grad_fn
y
```

```python
# 3rd option
with torch.no_grad():
    y = x + 2

y
```

Whenever we call the `backward()` function, the gradient for the tensor will be accumulated into the `dot_grad` attribute (summing up).

```python
weights = torch.ones(4, requires_grad=True)

for epoch in range(3):
    model_output =(weights * 3).sum()

    model_output.backward()

    print(weights.grad)
```

To avoid this, we need to empty the gradients with `grad.zero_()` after each iteration.

```python
weights = torch.ones(4, requires_grad=True)

for epoch in range(3):
    model_output =(weights * 3).sum()

    model_output.backward()

    print(weights.grad)

    weights.grad.zero_()
```

## Gradient Descent using Autograd

### From Scratch

- Prediction: Manually
- Gradient Computation: Manually
- Loss Computation: Manually
- Parameter Updates: Manually

```python
import numpy as np

# input & target
X = np.array([1, 2, 3, 4], dtype=np.float32)
y = np.array([2, 4, 6, 8], dtype=np.float32)

# weight initialization
w = 0.0

# model prediction
def forward(x):
    return w * x

# loss - MSE
def loss(y, y_pred):
    return ((y_pred - y) ** 2).mean()

# calculate gradient
def gradient(x, y, y_pred):
    return np.dot(2 * x, y_pred - y).mean()

print(f"Prediction before training: f(5) = {forward(5)}")
```

```python
# Training
learning_rate = 0.01
n_iters = 10

for epoch in range(n_iters):
    # prediction / forward pass
    y_pred = forward(X)

    # loss
    l = loss(y, y_pred)

    # gradients
    dw = gradient(X, y, y_pred)

    # update weights
    w -= learning_rate * dw

    if epoch % 1 == 0:
        print(f"epoch {epoch+1}: w = {w}, loss = {l}")

print(f"Prediction after training: f(5) = {forward(5)}")
```

### With PyTorch Gradients

- Prediction: Manually
- Gradient Computation: PyTorch
- Loss Computation: Manually
- Parameter Updates: Manually

```python
# input & target
X = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
y = torch.tensor([2, 4, 6, 8], dtype=torch.float32)

# weight initialization
w = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)

# model prediction
def forward(x):
    return w * x

# loss - MSE
def loss(y, y_pred):
    return ((y_pred - y) ** 2).mean()

print(f"Prediction before training: f(5) = {forward(5)}")

# Training
learning_rate = 0.01
n_iters = 100

for epoch in range(n_iters):
    # prediction / forward pass
    y_pred = forward(X)

    # loss
    l = loss(y, y_pred)

    # gradients = backward pass using PyTorch
    l.backward()

    # update weights - with PyTorch gradients
    # need to disable gradients
    with torch.no_grad():
        w -= learning_rate * w.grad

    # empty the gradients after each epoch not to accumulate
    w.grad.zero_()

    if epoch % 10 == 9:
        print(f"epoch {epoch+1}: w = {w}, loss = {l}")

print(f"Prediction after training: f(5) = {forward(5)}")
```

## Training Pipeline

There are generally 3 steps.

1. Designing our model(input size, output size, forward pass)
2. Constructing the loss and optimizer
3. Training Loop
    - Forward Pass - compute prediction
    - Backward Pass - gradients
    - Update Parameters

### Using PyTorch Autograd, Loss & Optimizer

```python
import torch
import torch.nn as nn

# input & target
X = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
y = torch.tensor([2, 4, 6, 8], dtype=torch.float32)

# weight initialization
w = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)

# model prediction
def forward(x):
    return w * x

print(f"Prediction before training: f(5) = {forward(5)}")

# Training
learning_rate = 0.01
n_iters = 100

# loss function using PyTorch NN
loss = nn.MSELoss()

# optimizer using PyTorch NN
optimizer = torch.optim.SGD([w], lr=learning_rate)

for epoch in range(n_iters):
    # prediction / forward pass
    y_pred = forward(X)

    # loss - the loss defined with PyTorch NN is a function
    l = loss(y, y_pred)

    # gradients = backward pass using PyTorch
    l.backward()

    # update weights - with PyTorch optimizer
    # no need to explicitly update anymore
    optimizer.step()

    # empty the gradients after each epoch not to accumulate
    optimizer.zero_grad()

    if epoch % 10 == 9:
        print(f"epoch {epoch+1}: w = {w}, loss = {l}")

print(f"Prediction after training: f(5) = {forward(5)}")
```

### Using PyTorch Model, Autograd, Loss & Optimizer

```python
# input & target - needs to be 2D matrix with PyTorch model
X = torch.tensor([[1], [2], [3], [4]], dtype=torch.float32)
y = torch.tensor([[2], [4], [6], [8]], dtype=torch.float32)

X_test = torch.tensor([5], dtype=torch.float32) # we also need a tensor as input to model for prediction

n_samples, n_features = X.shape
print(n_samples, n_features)

input_size = n_features
output_size = n_features
model = nn.Linear(input_size, output_size) # doesn't need weight initialization anymore

# model prediction
def forward(x):
    return w * x

# for prediction, we can now just call the model
print(f"Prediction before training: f(5) = {model(X_test).item()}")

# Training
learning_rate = 0.01
n_iters = 100

# loss function using PyTorch NN
loss = nn.MSELoss()

# optimizer using PyTorch NN
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

for epoch in range(n_iters):
    # prediction / forward pass
    y_pred = forward(X)

    # loss - the loss defined with PyTorch NN is a function
    l = loss(y, y_pred)

    # gradients = backward pass using PyTorch
    l.backward()

    # update weights - with PyTorch optimizer
    # no need to explicitly update anymore
    optimizer.step()

    # empty the gradients after each epoch not to accumulate
    optimizer.zero_grad()

    if epoch % 10 == 9:
        [w, b] = model.parameters()
        print(f"epoch {epoch+1}: w = {w[0][0].item()}, loss = {l}")

print(f"Prediction after training: f(5) = {model(X_test).item()}")
```

```python
class LinearRegression(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(LinearRegression, self).__init__()

        #define layers
        self.lin = nn.Linear(input_dim, output_dim)

    def forward(self, x):
        return self.lin(x)

model = LinearRegression(input_size, output_size)
```
