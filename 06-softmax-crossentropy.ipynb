# Softmax and Crossentropy

## Softmax

```python
import torch
import torch.nn as nn
import numpy as np

def softmax(x):
    return np.exp(x) / np.sum(np.exp(x), axis=0)

x = np.array([2.0, 1.0, 0.1])
outputs = softmax(x)
print('Softmax numpy: ', outputs)
```

```python
# with torch
x = torch.tensor([2.0, 1.0, 0.1])
outputs = torch.softmax(x, dim=0)
print('Softmax torch: ', outputs)
```

## Cross-entropy

```python
def crossentropy(actual, predicted):
    loss = - np.sum(actual * np.log(predicted))
    return loss

y = np.array([1, 0, 0])

y_pred_good = np.array([0.7, 0.2, 0.1])
y_pred_bad = np.array([0.1, 0.3, 0.6])
l1 = crossentropy(y, y_pred_good)
l2 = crossentropy(y, y_pred_bad)

print(f"Loss1 numpy: {l1}")
print(f"Loss2 numpy: {l2}")
```

With PyTorch loss function:

- No softmax needed.
- Ground truth class labels - should not be one-hot encoded.
- The prediction contains logit scores (no softmax)

```python
# with torch

loss = nn.CrossEntropyLoss() # no softmax needed

# 3 samples -> 3 class labels
y = torch.tensor([0])

y_pred_good = torch.tensor([[0.7, 0.2, 0.1]])
y_pred_bad = torch.tensor([[0.1, 0.3, 0.6]])

l1 = loss(y_pred_good, y)
l2 = loss(y_pred_bad, y)

print(f"Loss1 torch: {l1}")
print(f"Loss2 torch: {l2}")
```

```python
_, prediction1 = torch.max(y_pred_good, 1)
_, prediction2 = torch.max(y_pred_bad, 1)
prediction1, prediction2, torch.max(y_pred_good, 1)
```

```python
# 3 samples -> 3 class labels
y = torch.tensor([2, 0, 1])

y_pred_good = torch.tensor([
    [0.1, 1.0, 2.1],
    [2.0, 1.0, 0.1],
    [0.2, 3.0, 0.1],
])

l1 = loss(y_pred_good, y)
print(f"Loss 1: {l1}")
```

## Binary Classification

```python
import torch
import torch.nn as nn

class NeuralNet1(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(NeuralNet1, self).__init__()
        self.linear1 = nn.Linear(input_size, hidden_size)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out = self.linear1(x)
        out = self.relu(out)
        out = self.linear2(out)

        y_pred = torch.sigmoid(out)
        return y_pred

model = NeuralNet1(input_size=28*28, hidden_size=5)
criterion = nn.BCELoss()
```

## Multiclass Classification

```python
import torch
import torch.nn as nn

class NeuralNet2(nn.Module):
    def __init__(self, input_size, hidden_size, num_classes):
        super(NeuralNet2, self).__init__()
        self.linear1 = nn.Linear(input_size, hidden_size)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        out = self.linear1(x)
        out = self.relu(out)
        out = self.linear2(out)
        # no softmax needed here -> handled by loss function
        return out

model = NeuralNet2(input_size=28*28, hidden_size=5, num_classes=3)
criterion = nn.CrossEntropyLoss()
```
